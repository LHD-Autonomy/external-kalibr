```python
import numpy as np
import matplotlib.pyplot as plt

# Your R_ic matrix
R_ic = np.array([[ 0.802069,    0.13989114, -0.58061673],
                 [ 0.16772363,  0.88029536,  0.44378921],
                 [ 0.57319639, -0.45333271,  0.68259457]])

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot IMU Frame (Identity Matrix) at origin
ax.quiver(0, 0, 0, 1, 0, 0, color='r', label='IMU X (Forward)')
ax.quiver(0, 0, 0, 0, 1, 0, color='g', label='IMU Y (Left)')
ax.quiver(0, 0, 0, 0, 0, 1, color='b', label='IMU Z (Up)')

# Plot Camera Frame (Columns of R_ic) at origin
ax.quiver(0, 0, 0, R_ic[0,0], R_ic[1,0], R_ic[2,0], color='r', linestyle='--', label='Cam X (Right)')
ax.quiver(0, 0, 0, R_ic[0,1], R_ic[1,1], R_ic[2,1], color='g', linestyle='--', label='Cam Y (Down)')
ax.quiver(0, 0, 0, R_ic[0,2], R_ic[1,2], R_ic[2,2], color='b', linestyle='--', label='Cam Z (Look)')

ax.set_xlim([-1, 1]); ax.set_ylim([-1, 1]); ax.set_zlim([-1, 1])
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
plt.legend()
plt.title("IMU Frame (Solid) vs Camera Frame (Dashed)")
plt.show()
```
